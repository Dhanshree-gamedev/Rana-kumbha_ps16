```python
from openpyxl import Workbook, load_workbook
import os
import random

file = "atm.xlsx"


# Create Excel file
if not os.path.exists(file):

    wb = Workbook()
    ws = wb.active
    ws.title = "ATM"

    ws.append(["Account No", "Name", "Address", "Phone No", "Balance"])

    wb.save(file)


def menu():

    print("\n========== ATM MACHINE ==========")
    print("1. Create Account")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Show Balance")
    print("5. Show Account Details")
    print("6. Exit")


def login():

    account_no = input("Enter your account number: ")

    wb = load_workbook(file)
    ws = wb["ATM"]

    for row in range(2, ws.max_row + 1):

        if str(ws.cell(row=row, column=1).value) == account_no:

            print("Login successful.")
            return row

    print("Account not found.")
    return None


def new_account():

    account_holder_name = input("Enter your name: ")
    address = input("Enter your address: ")
    phone_no = input("Enter your phone number: ")

    while True:

        try:
            amount = int(input("Enter initial balance: "))

            if amount < 0:
                print("Balance cannot be negative.")
            else:
                break

        except ValueError:
            print("Please enter a valid amount.")


    # Generate account number
    account_no = random.randint(100000, 999999)

    wb = load_workbook(file)
    ws = wb["ATM"]

    # Check that account number is unique
    while True:

        found = False

        for row in range(2, ws.max_row + 1):

            if ws.cell(row=row, column=1).value == account_no:
                found = True
                break

        if not found:
            break

        account_no = random.randint(100000, 999999)


    ws.append([
        account_no,
        account_holder_name,
        address,
        phone_no,
        amount
    ])

    wb.save(file)

    print("\nAccount created successfully!")
    print("Your Account Number:", account_no)
    print("Your Initial Balance:", amount)


def deposit():

    row = login()

    if row is None:
        return

    while True:

        try:
            amount = int(input("Enter money to be deposited: "))

            if amount <= 0:
                print("Enter a positive amount.")
            else:
                break

        except ValueError:
            print("Please enter a valid amount.")


    wb = load_workbook(file)
    ws = wb["ATM"]

    balance = ws.cell(row=row, column=5).value

    balance = balance + amount

    ws.cell(row=row, column=5).value = balance

    wb.save(file)

    print("\nMoney deposited successfully.")
    print("Deposited Amount:", amount)
    print("Current Balance:", balance)


def withdraw():

    row = login()

    if row is None:
        return

    while True:

        try:
            withdraw_amount = int(
                input("Enter money to withdraw: ")
            )

            if withdraw_amount <= 0:
                print("Enter a positive amount.")
            else:
                break

        except ValueError:
            print("Please enter a valid amount.")


    wb = load_workbook(file)
    ws = wb["ATM"]

    balance = ws.cell(row=row, column=5).value


    if withdraw_amount > balance:

        print("\nTransaction failed.")
        print("Insufficient balance.")
        print("Available Balance:", balance)

    else:

        balance = balance - withdraw_amount

        ws.cell(row=row, column=5).value = balance

        wb.save(file)

        print("\nPlease collect your cash.")
        print("Withdrawn Amount:", withdraw_amount)
        print("Remaining Balance:", balance)


def showbalance():

    row = login()

    if row is None:
        return

    wb = load_workbook(file)
    ws = wb["ATM"]

    balance = ws.cell(row=row, column=5).value

    print("\nCurrent Balance:", balance)


def account_details():

    row = login()

    if row is None:
        return

    wb = load_workbook(file)
    ws = wb["ATM"]

    account_no = ws.cell(row=row, column=1).value
    name = ws.cell(row=row, column=2).value
    address = ws.cell(row=row, column=3).value
    phone_no = ws.cell(row=row, column=4).value
    balance = ws.cell(row=row, column=5).value

    print("\n========== ACCOUNT DETAILS ==========")
    print("Account Number :", account_no)
    print("Name           :", name)
    print("Address        :", address)
    print("Phone Number   :", phone_no)
    print("Balance        :", balance)


# Main program

while True:

    menu()

    try:
        ch = int(input("Enter your choice: "))

    except ValueError:
        print("Please enter a valid number.")
        continue


    if ch == 1:

        new_account()


    elif ch == 2:

        deposit()


    elif ch == 3:

        withdraw()


    elif ch == 4:

        showbalance()


    elif ch == 5:

        account_details()


    elif ch == 6:

        print("\nThank you for using our ATM.")

        # Download Excel file in Google Colab
        try:
            from google.colab import files
            files.download("atm.xlsx")
        except:
            pass

        break


    else:

        print("Enter a valid choice.")
```
